# Clase 3 de agosto

1. Escribir una función `potencia_dos(n)` que compute $2^n$
2. Una función recursiva `potencia_dos_recursiva(n)`que haga lo anterior
3. Una función recursiva pero rapida `potencia_dos_recursiva1(n)`que resuelva el problema anterior

```python
def potencia_dos(n):
  '''Retorna 2^n'''
  resultado = 1
  for i in range(n):
    resultado *= 2
  return resultado

potencia_dos(4), potencia_dos(0)
```

```python
def potencia_dos_recursiva(n):
  if n == 0:
    return 1
  else:
    a = potencia_dos_recursiva(n-1)
    return 2*a
```

```python
for i in range(4):
  print(potencia_dos_recursiva(i), end = ' ')
```

```python
def potencia_dos_recursiva1(n):
  if n == 0:
    return 1
  else:
    potencia = potencia_dos_recursiva1(n//2)
    if n%2 == 0:
      return potencia*potencia
    else:
      return 2*potencia*potencia
```

```python
potencia_dos_recursiva1(3)
```

```python
def maximo_recursivo(L):
  '''Retorna el máximo de la lista de manera recursiva'''
  if len(L) == 1:
    return L[0]
  maximotentativo = maximo_recursivo(L[:-1])
  if maximotentativo > L[-1]:
    return maximotentativo
  else:
    return L[-1]

maximo_recursivo([1,2,3,4,5,.7,-10,14, 23])
```

```python
def maximo_recursivo_cantidad(L):
  '''Retorna el máximo y el número de apariciones de una lista de manera recursiva'''
  if len(L) == 1:
    return L[0], 1
  maximotentativo, apariciones = maximo_recursivo_cantidad(L[:-1])
  if maximotentativo > L[-1]:
    return maximotentativo, apariciones
  elif maximotentativo == L[-1]:
    return maximotentativo, apariciones +1
  elif maximotentativo < L[-1]:
    return L[-1], 1
```

```python
maximo_recursivo_cantidad([1,1,2,3,4, 4,9])
```
